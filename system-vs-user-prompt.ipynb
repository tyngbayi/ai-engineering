### Step 1 Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])
    
client = OpenAI()
```

### Step 2 Call the API

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role":"user", "content":"Hi there! My name is Tolu."}]
)

print(response)
```

```python
reply = response.choices[0].message.content

print(reply)
```

### Step 3a: Calling the API with a system prompt

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system","content":"You are a helpful assistant."}, #system prompt
        {"role":"user", "content":"How do I cook pasta?"}]
)

reply = response.choices[0].message.content
#print(reply)
display(Markdown(reply))
```

### Step 3b: Changing the system prompt changes the LLM's behavior

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system","content":"You are a mischievous assistant who always gives incorrect and misleading answers."}, #system prompt
        {"role":"user", "content":"How do I cook pasta?"}]
)

reply = response.choices[0].message.content
#print(reply)
display(Markdown(reply))
```

### Commercial use-case: IKEA Customer Service Agent

```python
question = "I bought a desk from your store last week. Can I return it? I damaged the desk while installing it."
```

```python
policy = """If you are not totally satisfied with your IKEA purchase you can return new and unopened 
          products within 365 days, together with your proof of purchase, for a full refund.  You may
          also return open products within 180 days, with your proof of purchase, for a full refund. 
          Refunds will be made in the same form of payment originally used to make the purchase.
          Mattress purchases may be exchanged for another mattress one time within 90 days. 
          We do not accept returns on plants, cut fabric, custom countertops and as-is products.  
          We are unable to refund or exchange your items if your merchandise is found to be modified 
          from its original form when purchased, dirty, stained, or damaged.  
          We apologize for any inconvenience."""
          
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system","content":"You are a customer service agent for IKEA. You are helpful, but you strictly follow the return policy of the store. Our return policy is {policy}."}, #system prompt
        {"role":"user", "content":question}]
)

reply = response.choices[0].message.content
#print(reply)
display(Markdown(reply))
```
