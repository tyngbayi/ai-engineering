### Step 1: Import Libraries & API keys

```python
import os
from openai import OpenAI
from pprint import pprint
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")
```

### Step 2: Basic UI Interface

```python
import gradio as gr
```

```python
def respond_basic(message, history):
    return f"I am crazy about AI! ***m:{message}, ***h:{history}"

gr.ChatInterface(fn=respond_basic).launch()
```

### Step 3: Gradio inbrowser

```python
def respond_basic(message, history):
    return f"I am crazy about AI! And I'm chatting with you in the browser."


gr.ChatInterface(fn=respond_basic).launch(inbrowser=True)
```

### Step 4: Gradio + AI

```python
def respond_ai(message, history):
    messages = [{"role": "system", "content": "You are a helpful assistant."}] + history + [{"role": "user", "content": message}]
    client = OpenAI()
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=messages
    )
    reply = response.choices[0].message.content
    return reply
    

gr.ChatInterface(fn=respond_ai).launch(inbrowser=True)   
```
